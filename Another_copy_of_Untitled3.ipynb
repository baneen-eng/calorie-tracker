```python
import pandas as pd
food = pd.read_csv('food.csv')
nut = pd.read_csv('nutrient.csv')
print(nut[nut['id'] == 1008])
```

```python
fn = pd.read_csv('food_nutrient.csv', usecols=['fdc_id', 'nutrient_id', 'amount'])
kcal = fn[fn['nutrient_id'] == 1008][['fdc_id', 'amount']]
kcal = kcal.rename(columns={'amount': 'kcal'})
foods = food.merge(kcal, on='fdc_id')
print(foods.shape)
foods[['description', 'kcal']].head(10)
```

```python
foods = foods.sort_values('kcal')
print(foods[['description', 'kcal']].head(3))
print(foods[['description', 'kcal']].tail(3))
```
